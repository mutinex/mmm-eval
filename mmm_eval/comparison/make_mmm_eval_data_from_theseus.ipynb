```python
import os
import pandas as pd

customer = 'freedom'
model_path = f'../campaign/cmp_media_pert/{customer}/base/'
output_path = f'../mmm_eval_data/{customer}/'

models = os.listdir(model_path)


for model in models:
    dataset_path = os.path.join(model_path, model, 'dataset_space')
    datasets_available = os.listdir(dataset_path)
    dates = pd.read_parquet(os.path.join(dataset_path, 'date_range.parquet'))
    all_data = []
    for dataset in datasets_available:
        data = pd.read_parquet(os.path.join(dataset_path, dataset))
        if dataset in ['events.parquet', 'holidays.parquet','externals.parquet', 'owend_media.parquet', 
                       'earned_media.parquet', 'trade.parquet', 'pricing.parquet']:        
            data = data.loc[dates['period_start']]
            all_data.append(data)
        if dataset == 'media_spend.parquet':
            data = data.loc[dates['period_start']]
            channels = data.columns.tolist()
            data.columns = [col if col.endswith('_brand') or col.endswith('_category') else col + '_product' for col in channels]
            all_data.append(data)
        if dataset == 'sales.parquet':
            data = data.loc[dates['period_start']]            
            data.columns = ['revenue']
            all_data.append(data)            
        if dataset == 'volume.parquet':
            data = data.loc[dates['period_start']]
            data.columns = ['quantity']
            all_data.append(data)
            
    all_data = pd.concat(all_data, axis = 1).reset_index()
    all_data.rename(columns = {'period_start': 'date'}, inplace = True)
    os.makedirs(output_path, exist_ok = True)
    all_data.to_parquet(os.path.join(output_path, f'{model}_data.parquet'))
  
    
```
